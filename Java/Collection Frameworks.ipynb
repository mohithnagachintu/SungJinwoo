# Iterable & Iterator

## Iterable<E> (root of all collections)

| Method | Time | Description |
|--------|------|-------------|
| `iterator()` | O(1) create | Returns `Iterator<E>` for traversal |
| `forEach(Consumer)` | O(n) | Lambda over each element (Java 8+) |
| `spliterator()` | O(1) create | Split-capable iterator for parallel streams |

## Iterator<E>

| Method | Time | Description |
|--------|------|-------------|
| `hasNext()` | O(1) | More elements? |
| `next()` | O(1) | Advance and return |
| `remove()` | O(1)* | Remove last returned (*impl-dependent) |
| `forEachRemaining(Consumer)` | O(n) remaining | Process rest |

## ListIterator<E> (List only)

| Method | Time | Description |
|--------|------|-------------|
| `hasPrevious()` / `previous()` | O(1) | Bidirectional |
| `nextIndex()` / `previousIndex()` | O(1) | Index tracking |
| `set(E)` | O(1)* | Replace last returned |
| `add(E)` | O(1)* | Insert at cursor (*ArrayList O(n)) |

**Fail-fast:** Structural modification during iteration (except `Iterator.remove()`) → `ConcurrentModificationException`.


````java
import java.util.*;
List<Integer> nums = new ArrayList<>(List.of(10, 20, 30));
nums.forEach(x -> System.out.print(x + " "));
System.out.println();
ListIterator<Integer> li = nums.listIterator(nums.size());
while (li.hasPrevious()) System.out.print(li.previous() + " ");
System.out.println();
````

##### Tasks — Iterable & Iterator
1. Print all elements with enhanced for, `iterator()`, and `forEach` lambda.
2. Use `ListIterator` to reverse-print a list in O(n) without extra structure.
3. Remove evens from `[1..10]` using `Iterator.remove()` during traversal.
4. **Big task — Implement manual fail-fast demo: modify list during for-each and catch the exception**


````java
import java.util.*;

// Task 1 — print with enhanced for, iterator(), and forEach
List<String> fruits = new ArrayList<>(List.of("apple", "banana", "cherry"));

System.out.println("=== Task 1: enhanced for ===");
for (String f : fruits) System.out.print(f + " ");
System.out.println();

System.out.println("=== Task 1: iterator() ===");
Iterator<String> it = fruits.iterator();
while (it.hasNext()) System.out.print(it.next() + " ");
System.out.println();

System.out.println("=== Task 1: forEach lambda ===");
fruits.forEach(f -> System.out.print(f + " "));
System.out.println();

// Task 2 — reverse-print with ListIterator in O(n), no extra structure
List<Integer> nums = new ArrayList<>(List.of(10, 20, 30, 40));
System.out.println("=== Task 2: reverse with ListIterator ===");
ListIterator<Integer> li = nums.listIterator(nums.size());
while (li.hasPrevious()) System.out.print(li.previous() + " ");
System.out.println();

// Task 3 — remove evens from [1..10] using Iterator.remove()
List<Integer> oneToTen = new ArrayList<>();
for (int i = 1; i <= 10; i++) oneToTen.add(i);
Iterator<Integer> rm = oneToTen.iterator();
while (rm.hasNext()) {
    if (rm.next() % 2 == 0) rm.remove();
}
System.out.println("=== Task 3: odds only === " + oneToTen);

// Task 4 — fail-fast: modify list during for-each, catch exception
List<Integer> fail = new ArrayList<>(List.of(1, 2, 3, 4, 5));
System.out.println("=== Task 4: fail-fast demo ===");
try {
    for (Integer x : fail) {
        if (x == 3) fail.add(99); // structural modification mid-iteration
        System.out.print(x + " ");
    }
    System.out.println();
} catch (ConcurrentModificationException e) {
    System.out.println();
    System.out.println("Caught: " + e.getClass().getSimpleName());
    System.out.println("List state: " + fail);
}
````

# Interface → Collection

## Collection<E> extends Iterable

The **shared contract** for List, Set, and Queue. Maps are **not** Collections.

### Query methods

| Method | Time | Description |
|--------|------|-------------|
| `size()` | O(1) | Number of elements |
| `isEmpty()` | O(1) | True if size == 0 |
| `contains(Object o)` | O(n) avg* | Membership test (*O(1) hash-based sets/maps) |
| `containsAll(Collection c)` | O(n·m) | All elements of c present |

### Modify methods

| Method | Time | Description |
|--------|------|-------------|
| `add(E e)` | O(1)* | Insert element (*impl-dependent) |
| `remove(Object o)` | O(n)* | Remove first equal element |
| `addAll(Collection c)` | O(m)* | Add all from c |
| `removeAll(Collection c)` | O(n·m) | Remove all matching c |
| `retainAll(Collection c)` | O(n·m) | Keep only elements in c |
| `clear()` | O(n) | Remove all |

### Bulk & array

| Method | Time | Description |
|--------|------|-------------|
| `toArray()` | O(n) | Object[] copy |
| `toArray(T[] a)` | O(n) | Typed array copy |
| `stream()` | O(1) create | Sequential stream (Java 8+) |
| `parallelStream()` | O(1) create | Parallel stream |

### JDK 8+ defaults on Collection

| Method | Use |
|--------|-----|
| `removeIf(Predicate)` | Conditional bulk remove |
| `forEach(Consumer)` | Iterate with lambda |

**DSA note:** Choose implementation first; these methods behave the same API-wise but differ in complexity.

# List → LinkedList

## LinkedList<E>

Doubly linked list. Implements **List**, **Deque**, and **Queue**.

### LinkedList as List

| Method | Time | Description |
|--------|------|-------------|
| `get(i) / set(i,e)` | O(n) | Walk from nearer end |
| `add(e) / add(i,e)` | O(1) ends / O(n) middle | Append O(1) |
| `remove(i) / remove(o)` | O(n) | Find + unlink |
| `indexOf / contains` | O(n) | Linear |
### LinkedList as Deque/Queue (also in Deque section)

| Method | Time | Description |
|--------|------|-------------|
| `addFirst / addLast` | O(1) | Insert at ends |
| `offerFirst / offerLast` | O(1) | Queue API at ends |
| `removeFirst / removeLast` | O(1) | Remove from ends |
| `pollFirst / pollLast` | O(1) | Remove or null |
| `peekFirst / peekLast` | O(1) | View end without remove |
| `push / pop` | O(1) | Stack at front |
| `peek / poll / offer` | O(1) | Queue at ends |
| `getFirst / getLast` | O(1) | Throws if empty |

**Space:** O(n) with extra pointer overhead per node.

### When to use
Deque operations when `ArrayDeque` is unavailable; educational LRU; rarely for index-based loops.


````java
import java.util.*;
LinkedList<Integer> ll = new LinkedList<>(List.of(1, 2, 3));
ll.addFirst(0); ll.addLast(4);
System.out.println("deque: " + ll);
System.out.println("poll: " + ll.poll());
````
