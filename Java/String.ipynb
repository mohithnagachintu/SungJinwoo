# Overview

## String ecosystem in Java

```mermaid
classDiagram
    direction TB
    CharSequence <|.. String
    CharSequence <|.. StringBuilder
    CharSequence <|.. StringBuffer
    Appendable <|.. StringBuilder
    Appendable <|.. StringBuffer
    String : immutable
    StringBuilder : mutable, not thread-safe
    StringBuffer : mutable, synchronized
    StringTokenizer : legacy tokenizer
```

| Type | Mutable? | Thread-safe? | When to use |
|------|----------|--------------|-------------|
| `String` | No | Yes (immutable) | Literals, comparison, substring, return values |
| `StringBuilder` | Yes | No | Building strings in loops (DSA default) |
| `StringBuffer` | Yes | Yes | Legacy; avoid unless sync needed |
| `char[]` | Yes | No | In-place char manipulation, sensitive data |
| `StringTokenizer` | — | No | Legacy split; prefer `String.split` or regex |

### Golden rules for DSA

1. **`+` in loops on String = O(n²)** — use `StringBuilder`.
2. **`substring` (Java 7+)** copies char range — O(k) time and space for length k.
3. **Compare content** with `equals`, never `==` (except interned literals).
4. **To char array:** `s.toCharArray()` when you need in-place edits on a copy.


# Character Class

## `Character` — wrapper for primitive `char`

- `char` is a **primitive** (16-bit UTF-16 code unit); `Character` is the **wrapper class**.
- Autoboxing: `Character c = 'A';` — char auto-boxes to `Character`.
- Most day-to-day use: **static methods on `Character`** operating on a `char` value (no object needed).

### Primitive vs wrapper

| | `char` | `Character` |
|---|--------|-------------|
| Type | primitive | object (reference) |
| Default | `\u0000` | `null` |
| Methods | none | many (via static helpers + instance) |
| Collections | can't store directly | `List<Character>` |

### Key static methods (most used in DSA)

| Method | Description | Example |
|--------|-------------|---------|
| `isLetter(c)` | Unicode letter | `'A'` → true |
| `isDigit(c)` | Unicode digit | `'5'` → true |
| `isLetterOrDigit(c)` | letter or digit | `'x'` → true |
| `isAlphabetic(c)` | alphabetic (broader) | |
| `isUpperCase(c)` / `isLowerCase(c)` | case check | |
| `toUpperCase(c)` / `toLowerCase(c)` | case convert | `'a'` → `'A'` |
| `isWhitespace(c)` | space, tab, newline | |
| `isSpaceChar(c)` | space-like chars | |
| `compare(c1, c2)` | compare two chars | returns int like `compareTo` |
| `valueOf(c)` | char → `Character` | boxing |
| `toString(c)` | char → `String` | `"A"` |

### ASCII-range checks (common in LeetCode)

When you only care about `a-z`, `A-Z`, `0-9`:

```java
boolean lower = c >= 'a' && c <= 'z';
boolean upper = c >= 'A' && c <= 'Z';
boolean digit = c >= '0' && c <= '9';
```

**Note:** `Character.isDigit('²')` can be true for some Unicode digits. For strict `0-9`, use the ASCII range.

### `char` ↔ `String`

| Direction | Code |
|-----------|------|
| `char` → `String` | `Character.toString(c)` or `String.valueOf(c)` |
| `String` → `char` | `s.charAt(0)` (single char) |
| `String` → `char[]` | `s.toCharArray()` |
| `char[]` → `String` | `new String(arr)` or `String.valueOf(arr)` |

### Python vs Java

| Python | Java |
|--------|------|
| `c.isalpha()` | `Character.isLetter(c)` |
| `c.isdigit()` | `Character.isDigit(c)` |
| `c.isalnum()` | `Character.isLetterOrDigit(c)` |
| `c.upper()` / `c.lower()` | `Character.toUpperCase(c)` / `toLowerCase(c)` |
| `ord(c)` / `chr(n)` | `(int)c` / `(char)n` |

````java
// Tasks 4-6 — Character Class
boolean isPalindromeAlphaNum(String s) {
    int l = 0, r = s.length() - 1;
    while (l < r) {
        while (l < r && !Character.isLetterOrDigit(s.charAt(l))) l++;
        while (l < r && !Character.isLetterOrDigit(s.charAt(r))) r--;
        if (Character.toLowerCase(s.charAt(l)) != Character.toLowerCase(s.charAt(r))) return false;
        l++; r--;
    }
    return true;
}
String reverseOnlyLetters2(String s) {
    char[] a = s.toCharArray();
    int l = 0, r = a.length - 1;
    while (l < r) {
        if (!Character.isLetter(a[l])) { l++; continue; }
        if (!Character.isLetter(a[r])) { r--; continue; }
        char t = a[l]; a[l] = a[r]; a[r] = t; l++; r--;
    }
    return new String(a);
}
boolean isValidUsername(String u) {
    if (u == null || u.length() < 3 || u.length() > 20) return false;
    if (!Character.isLetter(u.charAt(0))) return false;
    for (char c : u.toCharArray())
        if (!(Character.isLetterOrDigit(c) || c == '_')) return false;
    return true;
}
System.out.println(isPalindromeAlphaNum("A man, a plan, a canal: Panama"));
System.out.println(reverseOnlyLetters2("a-bC-dEf-ghIj"));
System.out.println("valid user: " + isValidUsername("alice_99"));
System.out.println("bad user: " + isValidUsername("9bad"));
````

# Integer Class

## `Integer` — wrapper for primitive `int`

- `int` is a **32-bit primitive**; `Integer` is the **wrapper class** (immutable object).
- Autoboxing / unboxing happen automatically: `Integer x = 42;` and `int y = x;`
- **`Integer` is immutable** — like `String`, every "change" creates a new object.

### Primitive vs wrapper

| | `int` | `Integer` |
|---|-------|-----------|
| Type | primitive | object |
| Default | `0` | `null` |
| Null-safe | always has value | can be `null` |
| Collections | can't use directly | `List<Integer>`, `Map<K,Integer>` |
| Compare | `==` compares values | `==` compares references! use `equals` |

### Key methods (most used in DSA & backend)

| Method | Description | Example |
|--------|-------------|---------|
| `parseInt(String s)` | String → int | `"42"` → 42 |
| `parseInt(String s, int radix)` | parse base 2–36 | `"ff", 16` → 255 |
| `valueOf(int i)` | int → `Integer` | boxing |
| `valueOf(String s)` | String → `Integer` | |
| `toString(int i)` | int → String | `42` → `"42"` |
| `toBinaryString(i)` | binary repr | `5` → `"101"` |
| `toHexString(i)` | hex repr | `255` → `"ff"` |
| `compare(int a, int b)` | compare primitives | -1, 0, 1 |
| `compareTo(other)` | instance method | |
| `equals(Object)` | value equality | use instead of `==` |
| `intValue()` | unbox to int | |
| `MAX_VALUE` / `MIN_VALUE` | constants | ±2³¹−1 |
| `sum(a, b)` / `max(a, b)` / `min(a, b)` | Java 8+ helpers | |

### Parsing gotchas

```java
Integer.parseInt("42");     // 42
Integer.parseInt(" 42 ");   // NumberFormatException (no trim!)
Integer.parseInt("0xFF");   // NumberFormatException (use radix 16)
Integer.parseInt("FF", 16); // 255
Integer.valueOf("42");      // returns Integer object
```

### `==` vs `equals` (Integer cache)

Java caches `Integer` objects for **-128 to 127**:

```java
Integer a = 127, b = 127;
a == b;        // true (same cached object)

Integer c = 128, d = 128;
c == d;        // false (different objects)
c.equals(d);   // true (same value)
```

**Rule:** always use `equals()` or `compareTo()` for `Integer` objects; use `==` only for primitives.

### Useful conversions

| Task | Code |
|------|------|
| String → int | `Integer.parseInt(s)` |
| int → String | `Integer.toString(n)` or `String.valueOf(n)` |
| int → binary string | `Integer.toBinaryString(n)` |
| char digit → int | `c - '0'` (when `Character.isDigit(c)`) |
| int → char digit | `(char)('0' + n)` (when 0–9) |

### Python vs Java

| Python | Java |
|--------|------|
| `int("42")` | `Integer.parseInt("42")` |
| `str(42)` | `Integer.toString(42)` |
| `ord('A')` | `(int)'A'` |
| no cache surprise | watch Integer -128..127 cache |

# String (Immutable)

## String fundamentals

- Stored in **UTF-16** (`char` is 16-bit; supplementary chars use two `char`s — be aware for emoji).
- **String pool:** Literals like `"hello"` may be interned in heap pool; `intern()` adds runtime strings to pool.
- **Immutability:** Security (params, class names), thread safety, hash caching for `HashMap` keys.

### Key methods & complexity

| Method | Time | Space | Notes |
|--------|------|-------|-------|
| `length()` | O(1) | — | |
| `charAt(i)` | O(1) | — | |
| `substring(i,j)` | O(j-i) | O(j-i) | New string copy |
| `indexOf` / `lastIndexOf` | O(n·m) worst | — | KMP inside JDK |
| `equals` / `equalsIgnoreCase` | O(n) | — | |
| `compareTo` | O(n) | — | Lexicographic |
| `split(regex)` | O(n) avg | O(n) | Regex compile cost |
| `toCharArray()` | O(n) | O(n) | Copy |
| `+` concatenation | O(a+b) | O(a+b) | New string each time |

### Pros & Cons

| Pros | Cons |
|------|------|
| Immutable → safe as map keys | Every edit allocates new object |
| Rich API | `+` in loop is slow |
| Intern pool saves memory for literals | `substring` copies (Java 7+) |

````java
// Leetcode 917 - Reverse Only Letters
public String reverseOnlyLetters(String s) {
    char arr[] = s.toCharArray();
    int left = 0, right = arr.length - 1;
    while (left < right){
        if (!Character.isAlphabetic(arr[left])){
            left++; continue;
        } else if (!Character.isAlphabetic(arr[right])){
            right--; continue;
        } else {
            char temp = arr[right];
            arr[right] = arr[left];
            arr[left] = temp;
            right--; left++; 
        }
    }
    return new String(arr);
}
reverseOnlyLetters("a-bC-dEf-ghIj");
````

````java
// Leetcode 125 - Valid Palindrome
public boolean isPalindrome(String s) {
    int left = 0, right = s.length() - 1;
    while (left < right) {
        while (left < right && !Character.isLetterOrDigit(s.charAt(left)))
            left++;
        while (left < right && !Character.isLetterOrDigit(s.charAt(right)))
            right--;
        if (Character.toLowerCase(s.charAt(left)) != Character.toLowerCase(s.charAt(right))){
            return false;
        }
        left++;right--;
    }       
    return true;
}
isPalindrome("A man, a plan, a canal: Panama");
````

````java
// Leetcode 647 - Palindromic Substrings
public int countPalindromesExpandCenters(String s, int left, int right){
    int ans = 0;
    while (left >= 0 && right < s.length()){
        if (s.charAt(left) != s.charAt(right)){
            break;
        }
        left--;
        right++;
        ans++;
    }
    return ans;
}
public int countSubstrings(String s) {
    int n = s.length(), ans = 0;
    // Expand from Centers
    for (int i=0;i<n;i++){
        // odd - length palindromes
        ans += countPalindromesExpandCenters(s,i,i);
        // even - length palindromes
        ans += countPalindromesExpandCenters(s,i,i+1);   
    }
    return ans;
}
countSubstrings("abc");
````

````java
// Leetcode 443 - String Compression
public int compress(char[] chars) {
    int n = chars.length;
    StringBuilder sb = new StringBuilder();
    for (int i = 0; i < n; i++){
        int count = 1;
        sb.append(chars[i]);
        while (i + 1 < n && chars[i] == chars[i+1]){
            count++;
            i++;
        }
        if (count > 1){
            sb.append(count);
        }
    }
    for (int i=0;i<sb.length();i++){
        chars[i] = sb.charAt(i);
    }
    return sb.length();
}
compress(new char[]{'a','a','b','b','c','c','c'});
// But TC - O(N), SC - O(N) -> Not using in-place

// Leetcode 443 - String Compression (In-place using two pointers)      
public int compress2(char[] chars) {
    int n = chars.length;
    int index = 0;
    for (int i = 0; i < n; i++){
        int count = 1;
        while (i + 1 < n && chars[i] == chars[i+1]){
            count++;
            i++;
        }
        chars[index++] = chars[i];  
        if (count > 1){                
            String countStr = Integer.toString(count);
            for (int j = 0; j < countStr.length(); j++) {
                chars[index++] = countStr.charAt(j);
            }
        }       
    }
    return index;
}
compress2(new char[]{'a','a','b','b','c','c','c'});
````

````java
public String addStrings(String num1, String num2) {
    StringBuilder sb = new StringBuilder();
    int i = num1.length()-1, j = num2.length()-1, carry = 0;
    while (i >= 0 || j >= 0){
        int a = i >= 0 ? num1.charAt(i) - '0' : 0;
        int b = j >= 0 ? num2.charAt(j) - '0' : 0;
        int c = a + b + carry;
        if (c >= 10){ 
            carry = 1;
            c -= 10;
        } else {
            carry = 0;
        }
        sb.append((char)('0'+c));
        i--;j--;
    }
    if (carry > 0){
        sb.append((char)('0'+carry));
    }
    return sb.reverse().toString();
}
addStrings("999", "999");
````

````java
// StringBuilder — LC 394 Decode String, LC 6 ZigZag, LC 1047
import java.util.*;
String decodeString(String s) {
    Deque<StringBuilder> strSt = new ArrayDeque<>();
    Deque<Integer> numSt = new ArrayDeque<>();
    strSt.push(new StringBuilder()); int k = 0;
    for (char c : s.toCharArray()) {
        if (Character.isDigit(c)) k = k * 10 + (c - '0');
        else if (c == '[') { numSt.push(k); strSt.push(new StringBuilder()); k = 0; }
        else if (c == ']') {
            StringBuilder cur = strSt.pop();
            int rep = numSt.pop();
            for (int i = 0; i < rep; i++) strSt.peek().append(cur);
        } else strSt.peek().append(c);
    }
    return strSt.peek().toString();
}
String convertZigZag(String s, int rows) {
    if (rows == 1) return s;
    StringBuilder[] lines = new StringBuilder[rows];
    for (int i = 0; i < rows; i++) lines[i] = new StringBuilder();
    int r = 0, dir = 1;
    for (char c : s.toCharArray()) {
        lines[r].append(c);
        if (r == 0) dir = 1; else if (r == rows - 1) dir = -1;
        r += dir;
    }
    StringBuilder res = new StringBuilder();
    for (StringBuilder line : lines) res.append(line);
    return res.toString();
}
String removeDuplicates(String s) {
    StringBuilder sb = new StringBuilder();
    for (char c : s.toCharArray()) {
        if (sb.length() > 0 && sb.charAt(sb.length()-1) == c) sb.deleteCharAt(sb.length()-1);
        else sb.append(c);
    }
    return sb.toString();
}
System.out.println(decodeString("3[a]2[bc]"));
System.out.println(convertZigZag("PAYPALISHIRING", 3));
System.out.println(removeDuplicates("abbaca"));
````

````java
// StringBuilder — Calculator II + SQL IN + URL query
import java.util.*;
int calculateII(String s) {
    int n = s.length(), num = 0, last = 0, total = 0; char op = '+';
    for (int i = 0; i < n; i++) {
        char c = s.charAt(i);
        if (Character.isDigit(c)) num = num * 10 + (c - '0');
        if (!Character.isDigit(c) || i == n - 1) {
            if (op == '+') { total += last; last = num; }
            else if (op == '-') { total += last; last = -num; }
            else if (op == '*') last *= num;
            else if (op == '/') last /= num;
            op = c; num = 0;
        }
    }
    return total + last;
}
String sqlIn(String col, List<Integer> ids) {
    StringBuilder sb = new StringBuilder(col).append(" IN (");
    for (int i = 0; i < ids.size(); i++) {
        if (i > 0) sb.append(", ");
        sb.append(ids.get(i));
    }
    return sb.append(')').toString();
}
String toQuery(Map<String,String> params) {
    StringBuilder sb = new StringBuilder();
    int i = 0;
    for (var e : params.entrySet()) {
        if (i++ > 0) sb.append('&');
        sb.append(e.getKey()).append('=').append(e.getValue());
    }
    return sb.toString();
}
System.out.println(calculateII("3+2*2"));
System.out.println(sqlIn("user_id", List.of(1,2,5)));
System.out.println(toQuery(Map.of("page","2","size","50")));
````

````java
// StringTokenizer — shell-like tokenizer (respect quotes)
import java.util.*;
List<String> tokenizeCommand(String input) {
    List<String> tokens = new ArrayList<>();
    StringBuilder cur = new StringBuilder();
    boolean inQuote = false;
    for (int i = 0; i < input.length(); i++) {
        char c = input.charAt(i);
        if (c == '"') inQuote = !inQuote;
        else if (Character.isWhitespace(c) && !inQuote) {
            if (cur.length() > 0) { tokens.add(cur.toString()); cur.setLength(0); }
        } else cur.append(c);
    }
    if (cur.length() > 0) tokens.add(cur.toString());
    return tokens;
}
System.out.println(tokenizeCommand("cmd arg1 "quoted arg" arg2"));
````

````java
// StringTokenizer — expression lexer (NUMBER, OP, PAREN)
import java.util.*;
enum TokType { NUMBER, OP, LPAREN, RPAREN }
record Token(TokType type, String value) {}
List<Token> lexExpr(String s) {
    List<Token> out = new ArrayList<>();
    int i = 0;
    while (i < s.length()) {
        char c = s.charAt(i);
        if (Character.isWhitespace(c)) { i++; continue; }
        if (Character.isDigit(c)) {
            int j = i;
            while (j < s.length() && Character.isDigit(s.charAt(j))) j++;
            out.add(new Token(TokType.NUMBER, s.substring(i, j)));
            i = j;
        } else if ("+-*/".indexOf(c) >= 0) { out.add(new Token(TokType.OP, String.valueOf(c))); i++; }
        else if (c == '(') { out.add(new Token(TokType.LPAREN, "(")); i++; }
        else if (c == ')') { out.add(new Token(TokType.RPAREN, ")")); i++; }
        else throw new IllegalArgumentException("bad char " + c);
    }
    return out;
}
System.out.println(lexExpr("(12 + 345) * 6"));
````

> Mix of fundamentals · LeetCode-style · real-world backend scenarios.

````java
// char[] — reverse in place, anagram check
char[] reverseInPlace(char[] a) {
    int l = 0, r = a.length - 1;
    while (l < r) { char t = a[l]; a[l++] = a[r]; a[r--] = t; }
    return a;
}
boolean isAnagramChars(String a, String b) {
    if (a.length() != b.length()) return false;
    int[] f = new int[26];
    for (int i = 0; i < a.length(); i++) { f[a.charAt(i)-'a']++; f[b.charAt(i)-'a']--; }
    for (int x : f) if (x != 0) return false;
    return true;
}
char[] demo = "abcd".toCharArray();
System.out.println(new String(reverseInPlace(demo)));
System.out.println(isAnagramChars("anagram", "nagaram"));
````

````java
// char[] — LC 76 Minimum Window Substring
import java.util.*;
String minWindow(String s, String t) {
    if (t.isEmpty()) return "";
    int[] need = new int[128], have = new int[128];
    for (char c : t.toCharArray()) need[c]++;
    int required = t.length(), formed = 0, bestLen = Integer.MAX_VALUE, bestL = 0;
    int l = 0;
    for (int r = 0; r < s.length(); r++) {
        char cr = s.charAt(r);
        if (need[cr] > 0) {
            have[cr]++;
            if (have[cr] <= need[cr]) formed++;
        }
        while (formed == required) {
            if (r - l + 1 < bestLen) { bestLen = r - l + 1; bestL = l; }
            char cl = s.charAt(l++);
            if (need[cl] > 0) { if (have[cl] <= need[cl]) formed--; have[cl]--; }
        }
    }
    return bestLen == Integer.MAX_VALUE ? "" : s.substring(bestL, bestL + bestLen);
}
System.out.println(minWindow("ADOBECODEBANC", "ABC"));
````

````java
// char[] — Rabin-Karp rolling hash substring search
int rabinKarp(String text, String pattern) {
    int n = text.length(), m = pattern.length();
    if (m == 0 || m > n) return -1;
    long base = 256, mod = 1_000_000_007;
    long ph = 0, th = 0, pow = 1;
    for (int i = 0; i < m; i++) {
        ph = (ph * base + pattern.charAt(i)) % mod;
        th = (th * base + text.charAt(i)) % mod;
        if (i < m - 1) pow = (pow * base) % mod;
    }
    for (int i = 0; i <= n - m; i++) {
        if (ph == th && text.substring(i, i + m).equals(pattern)) return i;
        if (i < n - m) th = ((th - text.charAt(i) * pow % mod + mod) * base + text.charAt(i + m)) % mod;
    }
    return -1;
}
System.out.println(rabinKarp("abracadabra", "cad"));
````

````java
// Regex — LC 44 Wildcard Matching (DP, no regex)
boolean isMatch(String s, String p) {
    int n = s.length(), m = p.length();
    boolean[][] dp = new boolean[n+1][m+1];
    dp[0][0] = true;
    for (int j = 1; j <= m; j++) if (p.charAt(j-1) == '*') dp[0][j] = dp[0][j-1];
    for (int i = 1; i <= n; i++) for (int j = 1; j <= m; j++) {
        char pc = p.charAt(j-1);
        if (pc == '*') dp[i][j] = dp[i][j-1] || dp[i-1][j];
        else if (pc == '?' || s.charAt(i-1) == pc) dp[i][j] = dp[i-1][j-1];
    }
    return dp[n][m];
}
System.out.println(isMatch("aa", "a"));      // false
System.out.println(isMatch("aa", "*"));      // true
System.out.println(isMatch("cb", "?a"));     // false
System.out.println(isMatch("adceb", "*a*b")); // true
````

##### Final Tasks — Capstone (pick 2+ or do all)

1. **Longest substring without repeating characters** — Sliding window + `HashSet` or freq array.
2. **Group anagrams** — `Map<String, List<String>>` with sorted key or count signature.
3. **Longest palindromic substring** — Expand around center or DP.

````java
// Capstone — LC 3 + LC 49
import java.util.*;
int lengthOfLongestSubstring(String s) {
    int[] last = new int[128];
    Arrays.fill(last, -1);
    int best = 0, l = 0;
    for (int r = 0; r < s.length(); r++) {
        char c = s.charAt(r);
        if (last[c] >= l) l = last[c] + 1;
        last[c] = r;
        best = Math.max(best, r - l + 1);
    }
    return best;
}
List<List<String>> groupAnagrams(String[] strs) {
    Map<String, List<String>> map = new HashMap<>();
    for (String w : strs) {
        char[] a = w.toCharArray(); Arrays.sort(a);
        String key = new String(a);
        map.computeIfAbsent(key, k -> new ArrayList<>()).add(w);
    }
    return new ArrayList<>(map.values());
}
System.out.println(lengthOfLongestSubstring("abcabcbb"));
System.out.println(groupAnagrams(new String[]{"eat","tea","tan","ate","nat","bat"}));
````

````java
// Capstone — LC 5 + LC 28
String longestPalindrome(String s) {
    int n = s.length(), bestL = 0, bestLen = 1;
    for (int center = 0; center < n; center++) {
        for (int len : new int[]{expand(s, center, center), expand(s, center, center+1)}) {
            if (len > bestLen) { bestLen = len; bestL = center - (len - 1) / 2; }
        }
    }
    return s.substring(bestL, bestL + bestLen);
}
int expand(String s, int l, int r) {
    while (l >= 0 && r < s.length() && s.charAt(l) == s.charAt(r)) { l--; r++; }
    return r - l - 1;
}
int strStr(String hay, String needle) {
    if (needle.isEmpty()) return 0;
    for (int i = 0; i <= hay.length() - needle.length(); i++)
        if (hay.startsWith(needle, i)) return i;
    return -1;
}
System.out.println(longestPalindrome("babad"));
System.out.println(strStr("sadbutsad", "sad"));
````

````java
// Capstone — LC 140 Word Break II + serialize/deserialize
import java.util.*;
List<String> wordBreakII(String s, Set<String> dict) {
    Map<Integer, List<String>> memo = new HashMap<>();
    return dfs(s, 0, dict, memo);
}
List<String> dfs(String s, int start, Set<String> dict, Map<Integer,List<String>> memo) {
    if (start == s.length()) return List.of("");
    if (memo.containsKey(start)) return memo.get(start);
    List<String> res = new ArrayList<>();
    for (int end = start + 1; end <= s.length(); end++) {
        String word = s.substring(start, end);
        if (!dict.contains(word)) continue;
        for (String tail : dfs(s, end, dict, memo))
            res.add(word + (tail.isEmpty() ? "" : " " + tail));
    }
    memo.put(start, res);
    return res;
}
String encode(List<String> strs) {
    StringBuilder sb = new StringBuilder();
    for (String s : strs) sb.append(s.length()).append('#').append(s);
    return sb.toString();
}
List<String> decode(String s) {
    List<String> out = new ArrayList<>();
    int i = 0;
    while (i < s.length()) {
        int j = s.indexOf('#', i);
        int len = Integer.parseInt(s.substring(i, j));
        i = j + 1;
        out.add(s.substring(i, i + len));
        i += len;
    }
    return out;
}
Set<String> dict = Set.of("cat","cats","and","sand","dog");
System.out.println(wordBreakII("catsanddog", dict));
List<String> original = List.of("hello", "world|pipe", "with#hash");
String enc = encode(original);
System.out.println("encoded: " + enc);
System.out.println("decoded: " + decode(enc));
````
