# Overview

## String ecosystem in Java

```mermaid
classDiagram
    direction TB
    CharSequence <|.. String
    CharSequence <|.. StringBuilder
    CharSequence <|.. StringBuffer
    Appendable <|.. StringBuilder
    Appendable <|.. StringBuffer
    String : immutable
    StringBuilder : mutable, not thread-safe
    StringBuffer : mutable, synchronized
    StringTokenizer : legacy tokenizer
```

| Type | Mutable? | Thread-safe? | When to use |
|------|----------|--------------|-------------|
| `String` | No | Yes (immutable) | Literals, comparison, substring, return values |
| `StringBuilder` | Yes | No | Building strings in loops (DSA default) |
| `StringBuffer` | Yes | Yes | Legacy; avoid unless sync needed |
| `char[]` | Yes | No | In-place char manipulation, sensitive data |
| `StringTokenizer` | — | No | Legacy split; prefer `String.split` or regex |

### Golden rules for DSA

1. **`+` in loops on String = O(n²)** — use `StringBuilder`.
2. **`substring` (Java 7+)** copies char range — O(k) time and space for length k.
3. **Compare content** with `equals`, never `==` (except interned literals).
4. **To char array:** `s.toCharArray()` when you need in-place edits on a copy.


# String (Immutable)

## String fundamentals

- Stored in **UTF-16** (`char` is 16-bit; supplementary chars use two `char`s — be aware for emoji).
- **String pool:** Literals like `"hello"` may be interned in heap pool; `intern()` adds runtime strings to pool.
- **Immutability:** Security (params, class names), thread safety, hash caching for `HashMap` keys.

### Key methods & complexity

| Method | Time | Space | Notes |
|--------|------|-------|-------|
| `length()` | O(1) | — | |
| `charAt(i)` | O(1) | — | |
| `substring(i,j)` | O(j-i) | O(j-i) | New string copy |
| `indexOf` / `lastIndexOf` | O(n·m) worst | — | KMP inside JDK |
| `equals` / `equalsIgnoreCase` | O(n) | — | |
| `compareTo` | O(n) | — | Lexicographic |
| `split(regex)` | O(n) avg | O(n) | Regex compile cost |
| `toCharArray()` | O(n) | O(n) | Copy |
| `+` concatenation | O(a+b) | O(a+b) | New string each time |

### Pros & Cons

| Pros | Cons |
|------|------|
| Immutable → safe as map keys | Every edit allocates new object |
| Rich API | `+` in loop is slow |
| Intern pool saves memory for literals | `substring` copies (Java 7+) |


````java
// Demo: String basics & immutability
String a = "hello";
String b = a;
a = a + " world";  // a points to NEW object; b still "hello"
System.out.println("a: " + a);
System.out.println("b: " + b);
System.out.println("length: " + a.length());
System.out.println("charAt(0): " + a.charAt(0));
````

# String Comparison & String Pool

## `==` vs `equals` vs `compareTo`

| Operator/Method | Compares | Use |
|-----------------|----------|-----|
| `==` | References | Only when you mean same object (interned literals) |
| `equals()` | Character content | Default for equality |
| `equalsIgnoreCase()` | Content, case insensitive | User input |
| `compareTo()` | Lex order | Sorting, ordering |

```mermaid
flowchart LR
    L["Literal abc"] --> P[String Pool]
    N["new String(abc)"] --> H[Heap object]
    P -.->|intern| H
```

**`intern()`:** Returns pooled canonical copy — saves memory if many duplicates; use sparingly in DSA.


````java
// Demo: == vs equals
String s1 = "java";
String s2 = "java";
String s3 = new String("java");
System.out.println("s1==s2: " + (s1 == s2));
System.out.println("s1==s3: " + (s1 == s3));
System.out.println("s1.equals(s3): " + s1.equals(s3));
System.out.println("compareTo: " + "abc".compareTo("abd"));
````

# String Methods for DSA

## High-frequency methods

| Pattern | API |

### `split` gotcha

`"a..b".split(".")` splits on **regex** — `.` matches any char. Escape: `split("\\.")`.


````java
// Demo: split, join, format
import java.util.*;

String csv = "apple,banana,cherry";
String[] parts = csv.split(",");
System.out.println("split: " + Arrays.toString(parts));
System.out.println("join: " + String.join(" | ", parts));
System.out.println(String.format("count=%d name=%s", 42, "test"));
````

# StringBuilder

## StringBuilder (mutable, not thread-safe)

**Default capacity:** 16 chars; grows by doubling + 2 when full.

| Operation | Amortized | Notes |
|-----------|-----------|-------|
| `append` | O(1)* | *Occasional O(n) copy on grow |
| `insert` / `delete` | O(n) | Shifts chars |
| `reverse` | O(n) | In-place on internal array |
| `toString()` | O(n) | Creates immutable String |
| `charAt` / `setCharAt` | O(1) | |

**Space:** O(n) internal char array (may have unused capacity).

### When to use

- Any loop building a string
- Reverse in place on builder
- Removing chars from end (`setLength`)
- Efficient concatenation in DP / backtracking path building


````java
// Demo: StringBuilder — avoid O(n²) concat
StringBuilder sb = new StringBuilder();
for (int i = 0; i < 5; i++) sb.append(i).append('-');
System.out.println(sb.toString());
sb.reverse();
System.out.println("reversed: " + sb);
````

# StringTokenizer (Legacy)

## StringTokenizer

Legacy class in `java.util` — splits by delimiters, **ignores empty tokens** between consecutive delimiters differently than `split`.

```java
StringTokenizer st = new StringTokenizer("a,,b", ",");
// tokens: "a", "b"  (empty between commas skipped)
```

| Approach | Prefer? |
|----------|---------|
| `String.split` | Yes — regex, familiar |
| `StringTokenizer` | Legacy code only |
| Manual index scan | Best control in contests |

Modern alternative: `Scanner` for stdin; `split` for known strings.


````java
// Demo: StringTokenizer vs split
import java.util.*;

String data = "one,two,,three";
StringTokenizer st = new StringTokenizer(data, ",");
List<String> tok = new ArrayList<>();
while (st.hasMoreTokens()) tok.add(st.nextToken());
System.out.println("Tokenizer: " + tok);
System.out.println("split: " + Arrays.toString(data.split(",")));
````

# char[] vs String

## Working with `char[]`

| | String | char[] |
|---|--------|--------|
| Mutable | No | Yes (array is mutable) |
| Security | Stays in pool until GC | Can zero out after use |
| DSA | Return type, keys | In-place transforms on copy |

```java
char[] c = s.toCharArray();
// two-pointer, swap, in-place reverse on c
String result = new String(c);
```

**Tip:** For anagram problems, use `int[26]` frequency instead of sorting when alphabet is fixed.


````java
// Demo: char[] two-pointer palindrome
char[] c = "racecar".toCharArray();
int l = 0, r = c.length - 1;
boolean ok = true;
while (l < r) {
    if (c[l++] != c[r--]) { ok = false; break; }
}
System.out.println("palindrome: " + ok);
````

# Regex Basics (Pattern & Matcher)

## Regular expressions for strings

| Class | Use |
|-------|-----|
| `String.matches(regex)` | Whole string match |
| `String.replaceAll(regex, repl)` | Replace all matches |
| `Pattern.compile` + `Matcher` | Reuse compiled pattern |

Common patterns for DSA:

| Pattern | Meaning |
|---------|---------|
| `\\d+` | One or more digits |
| `[a-zA-Z]+` | Letters |
| `\\s+` | Whitespace |
| `^` / `$` | Start / end of string |
| `\\.` | Literal dot (escaped) |

**Performance:** Compile once, reuse `Pattern` in loops.


````java
// Demo: regex validation
String ip = "192.168.1.1";
boolean ok = ip.matches("\\d{1,3}(\\.\\d{1,3}){3}");
System.out.println("ip pattern (basic): " + ok);

String words = "cat bat rat";
long count = words.split("\\s+").length;
System.out.println("word count: " + count);
````

# DSA Patterns — String Masterclass

## Pattern catalog

```mermaid
flowchart TB
    S[String Problem] --> P1[Two Pointers]
    S --> P2[Sliding Window]
    S --> P3[Hash Map Freq]
    S --> P4[Stack]
    S --> P5[StringBuilder Build]
    S --> P6[Trie / Prefix]
    P1 --> E1[Palindrome, Valid Palindrome]
    P2 --> E2[Min Window, Longest Substring K Distinct]
    P3 --> E3[Anagram, Ransom Note]
    P4 --> E4[Decode String, Remove K Digits]
    P5 --> E5[Construct from parts]
    P6 --> E6[Autocomplete, Word Search II]
```

| Pattern | Template idea |
|---------|----------------|
| Two pointers | `l=0, r=n-1` or both from 0 |
| Sliding window | Expand `r`, shrink `l` while invalid |
| Prefix / rolling hash | Rabin-Karp, compare substrings O(1) |
| Stack | Matching, nested structure |
| Sort + join | Anagram grouping signature |


##### Final Tasks — Capstone (pick 2+ or do all)

1. **Longest substring without repeating characters** — Sliding window + `HashSet` or freq array.
2. **Group anagrams** — `Map<String, List<String>>` with sorted key or count signature.
3. **Longest palindromic substring** — Expand around center or DP.
4. **Implement strStr()** — Find needle in haystack (KMP optional bonus).
5. **Big task — Word break II** — Return all sentence splits; backtracking + memo + `HashSet` dictionary.
6. **Big task — Serialize/deserialize** — Design encode/decode for list of strings with delimiter that works for any content (length-prefix encoding).
